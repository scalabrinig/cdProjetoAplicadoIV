# **Resumo**

Este projeto aborda o desafio do atraso informacional na identificação de tendências do mercado de trabalho formal, um obstáculo que limita a eficácia do planejamento de instituições de qualificação e centros de empregabilidade. Alinhado ao Objetivo de Desenvolvimento Sustentável 8 (Trabalho Decente e Crescimento Econômico), o estudo propõe o desenvolvimento de um sistema de alerta precoce focado na previsão do saldo mensal de vagas por setor econômico. A metodologia baseia-se na extração e no tratamento dos microdados do Novo Cadastro Geral de Empregados e Desempregados (Novo CAGED), seguidos pela aplicação de modelagem de séries temporais. Para o desenvolvimento preditivo, propõe-se inicialmente a comparação entre um modelo de referência (baseline sazonal ingênuo) e técnicas avançadas, como a família de modelos SARIMA/SARIMAX que permite a inclusão de variáveis exógenas como eventos atípicos e o algoritmo Prophet, escolhido por sua arquitetura computacional moderna e robusta a anomalias e quebras de tendência. Ressalta-se que, a depender das restrições de cronograma e dos desafios técnicos encontrados ao longo da execução, o escopo poderá ser ajustado para focar exclusivamente no método que apresentar maior viabilidade operacional. Como principal resultado, espera-se que o produto analítico gerado forneça projeções precisas de contração ou expansão setorial, viabilizando uma tomada de decisão proativa e fundamentada em dados para o encaminhamento profissional e a estruturação de políticas de qualificação.

# **Introdução**

A área de Ciência de Dados tem revolucionado a formulação de políticas públicas e a análise socioeconômica, permitindo a extração de inteligência a partir de grandes volumes de informações históricas. Este projeto insere-se na intersecção entre a Ciência de Dados, a Economia e a Gestão de Políticas Públicas, alinhando-se diretamente ao Objetivo de Desenvolvimento Sustentável 8 (ODS-8) da Agenda 2030 da Organização das Nações Unidas (ONU, 2015), que preconiza o "Trabalho Decente e Crescimento Econômico".

O escopo temático concentra-se na modelagem preditiva da dinâmica do mercado de trabalho formal brasileiro, utilizando técnicas de Séries Temporais para analisar fluxos de admissões e desligamentos.

A motivação para este projeto surge da assimetria e do atraso informacional na identificação de tendências de contração ou expansão do mercado de trabalho. Atualmente, a formulação de diretrizes por parte de instituições educacionais e centros de empregabilidade ocorre, majoritariamente, de forma reativa, baseando-se em relatórios retrospectivos. Essa lacuna gera um descompasso estrutural crônico: profissionais são frequentemente capacitados ou direcionados para setores em retração econômica, enquanto áreas em franca expansão sofrem com o déficit de mão de obra qualificada. Em um ecossistema econômico volátil, a oportunidade de atuação está justamente na capacidade de antecipar o comportamento do mercado.

Nesse contexto, justifica-se a relevância desta pesquisa pelas contribuições diretas que ela pode oferecer aos agentes que compõem o ecossistema de qualificação e empregabilidade. A principal contribuição esperada é a entrega de um produto analítico de software, configurado como um sistema de alerta precoce que traduza dados brutos em inteligência acionável. De modo prático, a solução favorecerá a alocação eficiente de recursos para três perfis principais: Escolas Técnicas e Organizações de Formação (alinhando currículos à demanda futura), Centros Públicos de Emprego, Núcleos de Carreiras e ONGs (refinando o encaminhamento de candidatos a setores em expansão) e Secretarias Municipais e Observatórios do Trabalho (planejando políticas públicas ativas e articulando mutirões de empregabilidade mais assertivos, baseados em dados).

## **2. Objetivos**

### 2.1. Objetivo Geral

Para alcançar esses benefícios, o objetivo geral deste projeto é desenvolver e implementar um produto analítico de previsão, fundamentado em técnicas de Séries Temporais, que atue como um **sistema de alerta precoce** para a dinâmica do emprego formal no Brasil, prevendo o saldo mensal de vagas (expansão ou retração) por setor econômico.

### 2.2. Objetivos Específicos

Para viabilizar a construção desta ferramenta preditiva e responder à pergunta de negócio central (*"Quais setores provavelmente reduzirão ou aumentarão a geração de empregos nos próximos meses?"*), foram traçados os seguintes objetivos específicos:

* **Extração e Tratamento de Dados (ETL):** Coletar, limpar e estruturar os dados granulares agregados e históricos do Novo CAGED.
* **Análise Exploratória (EDA):** Identificar padrões subjacentes no mercado de trabalho, decompondo as séries temporais para compreender tendências de longo prazo, sazonalidades e anomalias.
* **Modelagem e Avaliação Preditiva:** Treinar, calibrar e comparar algoritmos de previsão adotando inicialmente o modelo estatístico SARIMA (HYNDMAN; ATHANASOPOULOS, 2018) e o algoritmo Prophet (TAYLOR; LETHAM, 2018), avaliados por métricas robustas à presença de zeros e valores negativos (como MAE, RMSE, sMAPE e MASE).
* **Geração de Inteligência Acionável:** Traduzir os resultados matemáticos do modelo selecionado em informações visuais diretas, mitigando a assimetria de informações no ecossistema de empregabilidade.
