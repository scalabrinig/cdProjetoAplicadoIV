# **Resumo**

Este projeto aborda o desafio do atraso informacional na identificação de tendências do mercado de trabalho formal, um obstáculo que limita a eficácia do planejamento de instituições de qualificação e centros de empregabilidade. Alinhado ao Objetivo de Desenvolvimento Sustentável 8 (Trabalho Decente e Crescimento Econômico), o estudo propõe o desenvolvimento de um sistema de alerta precoce focado na previsão do saldo mensal de vagas por setor econômico. Ressalta-se que, a depender das restrições de cronograma e dos desafios técnicos encontrados ao longo da execução, o escopo poderá ser ajustado para focar exclusivamente no método que apresentar maior viabilidade operacional. Como principal resultado, espera-se que o produto analítico gerado forneça projeções precisas de contração ou expansão setorial, viabilizando uma tomada de decisão proativa e fundamentada em dados para o encaminhamento profissional e a estruturação de políticas de qualificação.

Palavras-chave: Séries temporais; Mercado de trabalho; Novo CAGED; Modelagem preditiva; Ciência de dados.

# **Introdução**
A área de Ciência de Dados tem revolucionado a formulação de políticas públicas e a análise socioeconômica, permitindo a extração de inteligência a partir de grandes volumes de informações históricas.

Este projeto insere-se na intersecção entre a Ciência de Dados, a Economia e a Gestão de Políticas Públicas, alinhando-se diretamente ao Objetivo de Desenvolvimento Sustentável 8 (ODS-8) da Agenda 2030 da Organização das Nações Unidas (ONU, 2015), que preconiza o "Trabalho Decente e Crescimento Econômico".

O escopo temático concentra-se na modelagem preditiva da dinâmica do mercado de trabalho formal brasileiro, utilizando técnicas de Séries Temporais para analisar fluxos de admissões e desligamentos.

O problema central analisado neste trabalho é a assimetria e o atraso informacional na identificação de tendências de contração ou expansão do mercado de trabalho. Atualmente, a formulação de diretrizes por parte de instituições educacionais e centros de empregabilidade ocorre, majoritariamente, de forma reativa, baseando-se em relatórios retrospectivos. Esse cenário gera um descompasso estrutural crônico: profissionais são frequentemente capacitados ou direcionados para setores em retração econômica, enquanto áreas em franca expansão sofrem com o déficit de mão de obra qualificada.

## **1. Motivações e Justificativas**

A relevância desta pesquisa reside na premissa de que o fomento ao crescimento econômico sustentado e a mitigação do desemprego dependem da capacidade de antecipar o comportamento do mercado. Em um ecossistema econômico volátil, compreender precocemente quais Grupos de Atividade Econômica irão gerar ou fechar postos de trabalho formal é um insumo estratégico indispensável para o planejamento de curto e médio prazo.

Este projeto proporciona benefícios diretos aos agentes que compõem o ecossistema de qualificação e empregabilidade, favorecendo a alocação eficiente de recursos materiais e humanos. De modo prático, a solução atende a três perfis principais:
* **Escolas Técnicas e Organizações de Formação:** Poderão fundamentar a escolha de quais oficinas ou cursos rápidos ofertar nos ciclos seguintes, alinhando os currículos à demanda futura projetada.
* **Centros Públicos de Emprego, Núcleos de Carreiras e ONGs:** Serão capazes de refinar o encaminhamento de candidatos, priorizando o contato com empresas de setores em expansão e evitando direcionar esforços para áreas com indicativo de demissões em massa.
* **Secretarias Municipais e Observatórios do Trabalho:** Terão subsídios para planejar políticas públicas ativas, cruzando os alertas do modelo com a realidade regional para articular mutirões de empregabilidade mais assertivos.

A principal contribuição esperada com este desenvolvimento é a entrega de um produto analítico de software, um sistema de alerta precoce que traduza dados brutos em inteligência acionável. Ao reduzir a defasagem informacional e identificar riscos antecipadamente, o projeto instrumentaliza a tomada de decisão humana (*data-driven*), substituindo o planejamento empírico por ações direcionadas, o que contribui efetivamente para a promoção do trabalho decente.

## **2. Objetivos**

### 2.1. Objetivo Geral
O objetivo principal deste projeto é desenvolver e implementar um produto analítico de previsão, fundamentado em técnicas de Séries Temporais, que atue como um **sistema de alerta precoce** para a dinâmica do emprego formal no Brasil. O intuito é que este modelo computacional sirva como uma ferramenta de suporte à tomada de decisão, prevendo o saldo mensal de vagas (expansão ou retração) por setor econômico, de modo a orientar ações de qualificação profissional, planejamento de carreiras e formulação de políticas públicas.

### 2.2. Objetivos Específicos
Para viabilizar a construção desta ferramenta preditiva e responder à pergunta de negócio central (*"Quais setores provavelmente reduzirão ou aumentarão a geração de empregos nos próximos meses?"*), foram traçados os seguintes objetivos específicos:

* **Extração e Tratamento de Dados (ETL):** Coletar, limpar e estruturar os dados granulares agregados e históricos do Novo CAGED, garantindo a qualidade e a granularidade (setor econômico) necessárias para a modelagem.
* **Análise Exploratória (EDA):** Identificar padrões subjacentes no mercado de trabalho, decompondo as séries temporais para compreender tendências de longo prazo, sazonalidades (ciclos anuais de contratação) e anomalias/ruídos.
* **Modelagem e Avaliação Preditiva:** Treinar, calibrar e comparar algoritmos de previsão adotando inicialmente o modelo estatístico SARIMA (HYNDMAN; ATHANASOPOULOS, 2018) e o algoritmo computacional Prophet (TAYLOR; LETHAM, 2018), para determinar qual abordagem apresenta o menor erro de previsão (valiados por métricas robustas à presença de zeros e valores negativos, como MAE, RMSE, sMAPE e MASE).
* **Geração de Inteligência Acionável:** Traduzir os resultados matemáticos do modelo selecionado em informações visuais e diretas, configurando o produto final não apenas como um exercício algorítmico, mas como um instrumento prático para mitigar a assimetria de informações no ecossistema de empregabilidade.




## **3. Descrição da Base de Dados**
A base de dados selecionada para este projeto provém do Novo Cadastro Geral de Empregados e Desempregados (Novo CAGED), instrumento que atua como o principal termômetro do mercado de trabalho formal, regido pela Consolidação das Leis do Trabalho (CLT), no Brasil.

**Indicação da fonte de dados:** Os dados são oficiais e de domínio público, gerados e disponibilizados pelo Ministério do Trabalho e Emprego (MTE, 2026) por meio do Programa de Disseminação de Estatísticas do Trabalho (PDET). Utiliza-se o conjunto de planilhas estatísticas consolidadas mensais fornecidas pelo ministério.

**Forma e período de coleta:** A coleta possui natureza administrativa e contínua, sendo proveniente das declarações obrigatórias enviadas mensalmente pelos empregadores de todo o país por meio do eSocial. O recorte temporal definido para o treinamento dos modelos de Séries Temporais abrange o período de janeiro de 2020 a julho de 2026.

**Informações amostradas (Variáveis Contínuas):**
* Estoque: Volume total de vínculos trabalhistas ativos no fim do período.
* Admissões: Número de novos contratos formais registrados no mês.
* Desligamentos: Número de encerramentos de contratos laborais no mês.
* Saldos: Diferença líquida entre admissões e desligamentos (esta será a variável-alvo ou target do nosso modelo preditivo).

**Estruturação e organização dos dados:** Em vez de lidar com gigabytes de microdados brutos no nível do trabalhador, optou-se por utilizar as planilhas consolidadas do Ministério. Isso mantém a granularidade mensal e setorial necessária para a modelagem sem sobrecarregar o processamento computacional local das máquinas do grupo. Utilizaremos duas visões principais: a Temporal (agregação mensal exigida pelos algoritmos preditivos) e a Setorial, segmentada pela Classificação Nacional de Atividades Econômicas (CNAE 2.0), para gerar alertas específicos de mercado.

**Limitações da Base:** Ressalta-se antecipadamente que os dados contemplam exclusivamente o mercado formal (CLT), estão sujeitos a revisões retroativas por parte do CAGED e, devido à natureza metodológica do Novo CAGED, a série não será concatenada com o histórico anterior a 2020 para evitar distorções estatísticas. Ademais, o escopo é estritamente preditivo, sem a intenção de inferir causalidade econômica.
