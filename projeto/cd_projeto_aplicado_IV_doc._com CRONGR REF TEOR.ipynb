## **1. O Mercado de Trabalho Formal e o Novo CAGED**
O monitoramento do emprego formal no Brasil é historicamente fundamentado nos registros administrativos consolidados pelo Ministério do Trabalho e Emprego por meio do Novo Cadastro Geral de Empregados e Desempregados (Novo CAGED), que integra informações do eSocial, do CAGED tradicional e do Empregador Web (Brasil, 2026). O saldo mensal de vagas reflete um registro administrativo sujeito a revisões metodológicas e variações sazonais (Brasil, 2026). A construção de um modelo preditivo para este indicador não substitui a divulgação oficial, mas atua como um sistema de alerta precoce para antecipar tendências de expansão ou retração setorial, subsidiando políticas ativas de emprego e o planejamento de instituições de qualificação profissional.

## **2. Conceitos e Decomposição de Séries Temporais**
Uma série temporal consiste em uma sequência de observações numéricas ordenadas cronologicamente e coletadas em intervalos regulares (Hyndman; Athanasopoulos, 2018). Sob a ótica clássica, a série pode ser decomposta em quatro componentes fundamentais (Silva; Peres; Boscarioli, 2016):
* **Tendência ($T_t$):** Direção de longo prazo da série (crescente, decrescente ou estável).
* **Sazonalidade ($S_t$):** Padrões de flutuação periódicos e regulares vinculados a ciclos fixos de calendário.
* **Ciclos ($C_t$):** Variações oscilatórias de médio/longo prazo associadas a ciclos macroeconômicos, sem periodicidade fixa.
* **Resíduo ou Ruído ($\epsilon_t$):** Variações aleatórias e irregulares não explicadas pelas demais componentes.

Adicionalmente, propriedades como estacionariedade (constância de média e variância ao longo do tempo) e autocorrelação orientam a escolha das técnicas de modelagem.

## **3. Trabalhos Correlacionados e Uso de Variáveis Exógenas**
Na literatura recente, abordagens de previsão econômica têm explorado o enriquecimento de modelos temporais com dados de alta frequência e indicadores de intenção. Em estudo correlato, Gama e Arruda (2024) demonstraram que a combinação de séries administrativas do CAGED com indicadores digitais (como tendências de busca em plataformas web) melhorou a acurácia de previsão fora da amostra em relação a modelos puramente univariados. Essa evidência fundamenta a viabilidade de incorporar variáveis exógenas no pipeline preditivo, mantendo o rigor na validação por janelas temporais.

## **4. Alternativas de Modelagem: Vantagens e Limitações**
Para a previsão do saldo mensal de vagas, três principais alternativas metodológicas foram avaliadas:
* **SARIMA (Seasonal Autoregressive Integrated Moving Average):** Captura autocorrelações lineares e sazonalidades de forma altamente interpretável (Hyndman; Athanasopoulos, 2018). *Vantagens:* Rigor estatístico consolidado para séries estáveis. *Limitações:* Alta sensibilidade a choques estruturais severos e rigidez diante de mudanças abruptas de regime.
* **Prophet:** Desenvolvido pelo Meta, baseia-se em modelos aditivos generalizados com tendências por partes e múltiplos componentes sazonais (Taylor; Letham, 2018). *Vantagens:* Robusto a dados ausentes, feriados e quebras de tendência, com parametrização flexível. *Limitações:* Requer cautela nos parâmetros de suavização para evitar extrapolações instáveis.
* **Modelos de Aprendizado de Máquina (XGBoost, Redes LSTM):** *Vantagens:* Elevada capacidade de mapear relações não lineares complexas. *Limitações:* Exigem grande volume de dados históricos, maior esforço em engenharia de atributos e apresentam menor interpretabilidade gerencial ("caixa preta").

Diante desse panorama, o projeto adotará a comparação estruturada entre uma linha de base sazonal ingênua (*baseline*), o SARIMA e o Prophet, avaliados por métricas robustas a zeros e valores negativos (MAE, RMSE, sMAPE e MASE), selecionando o modelo que melhor concilie acurácia estatística e utilidade prática para os gestores educacionais e públicos.

Decidiu-se que as séries de admissões e desligamentos do Novo CAGED serão analizadas em valores absolutos, ainda que, à primeira vista, possa parecer lógico que se considere o impacto do crescimento populacional em cada estado como uma fonte de distorção em cada série. ## Parâmetro de Box-Cox e método de Guerrero

````python
from ast_caged.modules.preprocessing.transformation import lambda_guerrero, box_cox, max_min
````

````python
# Método de Guerrero para os lambda de Box-Cox

for uf in time_series_uf:
    print(uf)

    time_series_uf[uf]["lmbda_origin"] = {}

    lmbda_admissoes = lambda_guerrero(time_series_uf[uf]["df"]["admissoes"])
    time_series_uf[uf]["lmbda_origin"]["admissoes"] = lmbda_admissoes
    
    lmbda_desligamentos = lambda_guerrero(time_series_uf[uf]["df"]["desligamentos"])
    time_series_uf[uf]["lmbda_origin"]["desligamentos"] = lmbda_desligamentos
````

````python
for uf in time_series_uf:
    print(uf,time_series_uf[uf]["lmbda_origin"])
````

Os resultados do método de guerrero sobre as séries até julho de 2026 mostram valores negativos para admissões (a escala deve ser comprimida) e positivos para desligamentos de Minas Gerais e São Paulo (a escala deve ser expandida). Valores próximos de 1 não exigiriam transformação Box-Cox (GUERRERO, 1993); entretanto, para evitar que este projeto se torne obsoleto com o tempo e reduzir necessidade de manutenção, sob a possibilidade de que essas necessidades de transformação se altere conforme mais meses são aderidos à série, e sob a premissa de que lambda próximo de 1 é despresível porque a transformação de Box-Cox surtiria efeitos despresíveis, automatizamos a transformação segundo o lambda armazenado sob a chave de cada estado.

````python
for uf in time_series_uf:

    serie_admi = time_series_uf[uf]["df"]["admissoes"]
    lmbda_admi = time_series_uf[uf]["lmbda_origin"]["admissoes"]
    transformed_admi = box_cox(serie_admi, lmbda_admi)
    time_series_uf[uf]["df"][transformed_admi.name] = transformed_admi

    serie_desl = time_series_uf[uf]["df"]["desligamentos"]
    lmbda_desl = time_series_uf[uf]["lmbda_origin"]["desligamentos"]
    transformed_desl = box_cox(serie_desl, lmbda_desl)
    time_series_uf[uf]["df"][transformed_desl.name] = transformed_desl
````

````python
# Plot comparativo

for title, df in time_series_uf.items():
    plot_series(df["df"][["admissoes", "admissoes_boxcox"]], title, "Competência", "Empregos")
    plot_series(df["df"][["desligamentos", "desligamentos_boxcox"]], title, "Competência", "Empregos")
````

````python
# Plot comparativo a despeito da amplitude

for title, df in time_series_uf.items():
    provisório_admi = pd.DataFrame()
    provisório_admi["admissoes"] = max_min(df["df"]["admissoes"])
    provisório_admi["admissoes_boxcox"] = max_min(df["df"]["admissoes_boxcox"])
    plot_series(provisório_admi[["admissoes", "admissoes_boxcox"]], title, "Competência", "Empregos")
    
    provisório_desl = pd.DataFrame()
    provisório_desl["desligamentos"] = max_min(df["df"]["desligamentos"])
    provisório_desl["desligamentos_boxcox"] = max_min(df["df"]["desligamentos_boxcox"])
    plot_series(provisório_desl[["desligamentos", "desligamentos_boxcox"]], title, "Competência", "Empregos")
````

Para saber se o efeito da transformação Box-Cox é útil para o modelo, a escolha da transformação será utilizada como um hiperparâmetro ajustável por busca (grid search ou random search).

*Adendo*: para lambda negativo, a escala de transformação possui um limite superior. Logo, uma previsão do modelo que ultrapasse esse limite não terá inversa Box-Cox válida. Isso precisa ser considerado quando em posse dos resultados finais.

## Testes de estacionariedade — ADF e KPSS

Antes de utilizar modelos de séries temporais, precisamos verificar se a série é estacionária.

De forma simples, uma série estacionária é aquela cujo comportamento estatístico permanece relativamente estável ao longo do tempo. Isso é importante porque modelos como o SARIMA dependem dessa característica ou de transformações que tornem a série estacionária.

Vamos utilizar dois testes: **Augmented Dickey-Fuller** (ADF) e **Kwiatkowski-Phillips-Schmidt-Shin** (KPSS)

### ADF — Augmented Dickey-Fuller

No teste ADF:

- **H0 (hipótese nula):** a série possui raiz unitária e não é estacionária;
- **H1 (hipótese alternativa):** a série é estacionária.

Se o p-valor for menor que 0,05, rejeitamos H0 e temos evidência de estacionariedade.

````python
from ast_caged.modules.inference.tests import adf_test, kpss_test
````

````python
# Testes de ADF

for uf in time_series_uf:

    time_series_uf[uf]["adf"] = {}

    adf_admissoes = adf_test(time_series_uf[uf]["df"]["admissoes"])
    time_series_uf[uf]["adf"]["admissoes"] = adf_admissoes
    
    adf_desligamentos = adf_test(time_series_uf[uf]["df"]["desligamentos"])
    time_series_uf[uf]["adf"]["desligamentos"] = adf_desligamentos

    print(uf)
    pprint(time_series_uf[uf]["adf"])
    print()
````

````python
for uf in time_series_uf:

    print(uf)
    print("      Admissões",time_series_uf[uf]["adf"]["admissoes"]["adf_pvalor"])
    print("  Desligamentos",time_series_uf[uf]["adf"]["desligamentos"]["adf_pvalor"])
    print()
````

### KPSS — Kwiatkowski-Phillips-Schmidt-Shin

O KPSS trabalha com uma lógica complementar:

- **H0:** a série é estacionária;
- **H1:** a série não é estacionária.

Se o p-valor for menor que 0,05, rejeitamos H0 e encontramos evidência de não estacionariedade.

Utilizar os dois testes ajuda a deixar o diagnóstico mais confiável.

### Análise e Interpretação dos Testes de Estacionaridade (ADF e KPSS)

A aplicação combinada dos testes **Augmented Dickey-Fuller (ADF)** e **Kwiatkowski-Phillips-Schmidt-Shin (KPSS)** sobre as 8 séries temporais individualizadas (Admissões e Desligamentos para as 4 UFs do Sudeste) fundamenta a necessidade metodológica de modelar os fluxos separadamente em vez de analisar um saldo agregado:

#### 1. Comportamento das Séries de Desligamentos (MG, ES, RJ e SP)
* **Diagnóstico Estatístico:** Em todas as quatro unidades federativas, a série de desligamentos apresentou **não-estacionaridade inequívoca**.
* **Validação dos Testes:**
  * O teste ADF obteve $p\text{-valor} > 0{,}05$ para todos os estados (MG: $0{,}0619$; ES: $0{,}3792$; RJ: $0{,}6618$; SP: $0{,}1930$), falhando em rejeitar a hipótese nula ($H_0$) de existência de raiz unitária.
  * Concomitantemente, o teste KPSS resultou em $p\text{-valor} \le 0{,}01$ (estatísticas de teste substancialmente superiores ao valor crítico de $0{,}463$ a $5\%$), rejeitando com segurança a hipótese nula ($H_0$) de estacionaridade em torno da média/nível.
* **Interpretação Econômica:** O fluxo de demissões acompanha tanto a expansão do estoque formal de empregos ao longo dos anos quanto uma forte sazonalidade determinística de final de ano (rescisões contratuais em novembro/dezembro). Essas séries exigirão **primeira diferença regular ($d = 1$)** e provável componente sazonal ($D = 1, s = 12$) na etapa de modelagem SARIMA.

#### 2. Comportamento das Séries de Admissões
* **Minas Gerais, Espírito Santo e Rio de Janeiro (Diagnóstico "Indeciso / Estacionária em Tendência"):**
  * O teste ADF indicou ausência de raiz unitária com significância estatística ($p\text{-valor} < 0{,}05$: MG com $0{,}0004$, ES com $0{,}0081$ e RJ com $0{,}0204$).
  * Contudo, o teste KPSS rejeitou a estacionaridade em nível ($p\text{-valor} \le 0{,}01$).
  * **Interpretação Econômica:** Este conflito clássico entre ADF e KPSS indica que as admissões desses estados não são puramente ruídos estacionários em torno de uma média fixa, mas sim séries **estacionárias em torno de uma tendência (*trend-stationary*)** ou sujeitas a quebras estruturais e ciclos sazonais marcantes.
* **São Paulo (Diagnóstico "Não-estacionária"):**
  * Em São Paulo, o teste ADF não rejeitou a raiz unitária ($p\text{-valor} = 0{,}2085$) e o KPSS rejeitou a estacionaridade ($p\text{-valor} = 0{,}01$). A dinâmica de contratação do maior polo econômico do país apresenta inércia e tendência estocástica de crescimento mais pronunciadas, necessitando de diferenciação ($d = 1$).

#### 3. Conclusão Metodológica para as Próximas Etapas
* **Inviabilidade do Saldo Direto:** A heterogeneidade entre as ordens de integração de admissões e desligamentos comprova que o saldo líquido mistura ordens estocásticas distintas, além de conter valores negativos que inviabilizam transformações de variância (como logaritmo).
* **Diretriz de Modelagem:** O pipeline prosseguirá com a modelagem independente das **8 séries de origem**. A projeção do saldo final de cada UF será obtida formalmente pela relação contábil entre as previsões individuais:
$$\widehat{\text{Saldo}}_{\text{UF}, t} = \widehat{\text{Admissões}}_{\text{UF}, t} - \widehat{\text{Desligamentos}}_{\text{UF}, t}$$
